```python
import numpy as np

class MonitorTemperatura:
    def __init__(self, limite_min=20.0, limite_max=80.0):
        self.leituras = []
        self.limite_min = limite_min
        self.limite_max = limite_max

    def adicionar_leitura(self, temperatura):
        #Adiciona uma nova leitura de temperatura (em °C).
        self.leituras.append(temperatura)
        print(f"Adiciona Leitura : {temperatura:.2f}°C")

    def calcular_estatisticas(self):
        #Calcula média, mínimo e máximo das temperaturas usando numpy.
        if not self.leituras:
            print("Nenhuma leitura disponível para análise.")
            return None

        arr = np.array(self.leituras)
        media = np.mean(arr)
        minimo = np.min(arr)
        maximo = np.max(arr)

        print("\n Estatísticas de Temperatura:")
        print(f"Temperatura média ..: {media:.2f}°C")
        print(f"Temperatura mínima .: {minimo:.2f}°C")
        print(f"Temperatura máxima .: {maximo:.2f}°C")

        return {"media": media, "min": minimo, "max": maximo}

    def verificar_segurança(self):
        #Verifica se as temperaturas estão dentro dos limites seguros.
        if not self.leituras:
            print("Nenhuma leitura registrada para verificação.")
            return

        print("\n Verificação de Segurança:")
        for temp in self.leituras:
            if temp < self.limite_min:
                print(f"ALERTA: Temperatura muito baixa ({temp:.2f}°C)! Abaixo de {self.limite_min}°C.")
            elif temp > self.limite_max:
                print(f"ALERTA: Temperatura muito alta ({temp:.2f}°C)! Acima de {self.limite_max}°C.")
            else:
                print(f"Temperatura dentro do limite seguro: {temp:.2f}°C")

    def resumo_geral(self):
        #Executa a análise completa: estatísticas + verificação de segurança.
        print("\n==============================")
        print("🌡️  MONITORAMENTO DE TEMPERATURA")
        print("==============================")
        self.calcular_estatisticas()
        self.verificar_segurança()
        print("==============================\n")


# Rotina Principal
if __name__ == "__main__":
    monitor = MonitorTemperatura()

    # Simulando inserção de leituras
    leituras_exemplo = [18.5, 22.3, 35.6, 78.9, 82.4, 65.2, 19.8, 25.0]
    for leitura in leituras_exemplo:
        monitor.adicionar_leitura(leitura)

    # Exibição das análises
    monitor.resumo_geral()
```
