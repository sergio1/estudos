```python
class Sensor:
    def __init__(self, localizacao, min_temp, max_temp):
        self.localizacao = localizacao
        self.min_temp = min_temp
        self.max_temp = max_temp
        self.operacional = True

    def calibrar(self):
        print(f"Sensor em {self.localizacao} calibrado.")

    def desligar(self):
        self.operacional = False
        print(f"Sensor em {self.localizacao} desligado.")

sensor1 = Sensor("Tanque A", -10, 60)
sensor2 = Sensor("Tanque B", 0, 50)
```
