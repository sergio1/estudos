```python
bebidas = ["Suco", "Chá", "", "Chá", "Chá gelado"]
bebidas_unicas = []

for item in bebidas: # Isso abrange todos os itens na estrutura bebidas

    if item not in bebidas_unicas: # Isso vai adicionar todo item de bebidas que não estiver em bebidas_unicas

        bebidas_unicas.append(item)
print(bebidas_unicas)
```

```python
manutencoes = ["motor", "sensor", "válvula", "motor"]
unicos = set(manutencoes)
print(unicos)

manutencoes = ["motor", "sensor", "válvula", "motor"]
unicos = list(manutencoes)
print(unicos)

manutencoes = ["motor", "sensor", "válvula", "motor"]
unicos = []
for item in manutencoes:
  if item not in unicos:
    unicos.append(item)
  print(unicos)
```
