```python
# 1-Importação das bibliotecas
import numpy as np

# 2-Definindo a classe MonitordeTemperadura
class MonitorDeTemperatura:
  '''
  classe para monitorar leituras de temperatura,
  calcular estatitisticas e verificar se estã em niveis seguros.
  (Docstring: documentação da classe)
  '''
  # Método construtor da classe
  def __init__(self):
    # Lista para armazenar as leitura de temperatura
    self.leituras = []

    # 3-Método para adicionar_leitura
  def adicionar_leitura(self, valor):
    '''
    adiciona uma nova leitura de temperatura à lista.
    '''
    # Adicionando o 'valor' a lista de leituras do objeto.
    self.leituras.append(valor)
    print(f"Nova Leitura Registrada : {valor} C")

  # 4-Método calcular_estatistica
  def calcular_estatistica(self):
    '''
    Calcula e retorna as estatisticas basicas das temperaturas.
    a média, mediana e desvio padrão
    '''
    # Converte a lista em leituras para array do NumPy.
    dados = np.array(self.leituras)

    # Cria um dicionario para armazenar os resultados das estatisticas.
    estatisticas = {
        "maximas": dados.max(),
        "minimas": dados.min(),
        "media": dados.mean(),
        "desvio padrao":dados.mean()
    }
    return estatisticas

  # 5-Método vetificar_seguranca
  def verificar_seguranca(self):
    '''
    verificar se todas as temperaturas estão dentro do intervalo segurao(20ºC a 80ºC).
    '''
    # Variavel booleana para rastrear se alguma leitura esta fora do limite
    fora_do_limite = False

    # Itere sobre cada temperatura atual está abaixo de 20C ou acima de 80C
    for temp in self.leituras:
      # Se estiver fora, imprime um alerta.
      if temp < 20 or temp > 80:
        print(f"Alerta: Teperatura fora do limite seguro detectada: {temp}ºC")
        fora_do_limite = True # seta a variavel para verdadeiro.

    # Se a vairavel 'fora_do_limite' ainda for falsa(ou seja, NUNHUMA temperatura estava fora do limite)
    if not fora_do_limite:
      print("Todas as temperatura estão dentro do limite seguro.")

# 6-Demonstrativo do uso do sistema
# criar o monitor de temperatura
monitor = MonitorDeTemperatura()

# Lista de temperaturas simuladas
leituras=[25,30,90,45,70]

# Adicionar cadas leitura ao monitor
for temp in leituras:
  monitor.adicionar_leitura(temp)

# Calculo e exibir as estatisticas
estatisticas = monitor.calcular_estatistica()
print("\nEstatisticas das Temperaturas Registradas:")
print(f"Temperatura Máxima: {estatisticas['maximas']}ºC")
print(f"Temperatura Mínima: {estatisticas['minimas']}ºC")
print(f"Temperatura Média: {estatisticas['media']}ºC")
print(f"Desvio Padrão: {estatisticas['desvio padrao']}ºC")

# Verificar se estão dentro do limite seguro
monitor.verificar_seguranca()
```

```python
# Importação das bibliotecas 📚
import numpy as np # Importa a biblioteca 'numpy', que é essencial para operações
                   # matemáticas eficientes em arrays e para calcular estatísticas
                   # como máximo, mínimo e média dos dados de temperatura.

# Definição da classe MonitorDeTemperatura com seu inicializador
class MonitorDeTemperatura:
    """
    Classe para monitorar leituras de temperatura,
    calcular estatísticas e verificar se estão em níveis seguros.
    (Docstring: documentação da classe)
    """

    # O método '__init__' é o construtor da classe. Ele é chamado
    # automaticamente ao criar um novo objeto (instância) da classe.
    def __init__(self):
        # Cria um atributo de instância chamado 'self.leituras'.
        # É uma lista vazia que será usada para armazenar todas as
        # leituras de temperatura adicionadas ao monitor.
        self.leituras = []

    # Método adicionar_leitura
    # Recebe um 'valor' (a nova temperatura) como argumento.
    def adicionar_leitura(self, valor):
        """
        Adiciona uma nova leitura de temperatura à lista.
        (Docstring: documentação do método)
        """
        # Adiciona o 'valor' à lista de leituras do objeto.
        self.leituras.append(valor)
        # Imprime uma confirmação de que a leitura foi registrada.
        print(f"Nova leitura registrada: {valor} °C")

    # Método calcular_estatisticas
    def calcular_estatisticas(self):
        """
        Calcula e retorna as estatísticas básicas das temperaturas.
        (Docstring: documentação do método)
        """
        # Converte a lista de leituras para um array do NumPy.
        # Isso permite usar as funções otimizadas do NumPy para cálculos.
        dados = np.array(self.leituras)

        # Cria um dicionário para armazenar os resultados das estatísticas.
        estatisticas = {
            # 'dados.max()' retorna o maior valor no array.
            "máxima": dados.max(),
            # 'dados.min()' retorna o menor valor no array.
            "mínima": dados.min(),
            # 'dados.mean()' retorna a média aritmética dos valores no array.
            "média": dados.mean()
        }
        # Retorna o dicionário de estatísticas.
        return estatisticas

    # Método verificar_seguranca
    def verificar_seguranca(self):
        """
        Verifica se todas as temperaturas estão dentro do intervalo seguro (20°C a 80°C).
        (Docstring: documentação do método)
        """
        # Flag booleana (indicador) para rastrear se alguma leitura
        # fora do limite foi encontrada. Começa como Falso.
        fora_do_limite = False

        # Itera sobre cada temperatura registrada na lista 'self.leituras'.
        for temp in self.leituras:
            # Verifica se a temperatura atual está abaixo de 20°C OU acima de 80°C.
            if temp < 20 or temp > 80:
                # Se estiver fora, imprime um ALERTA.
                print(f"‼️ Alerta: Temperatura fora do limite seguro detectada: {temp} °C")
                # Seta a flag para Verdadeiro.
                fora_do_limite = True

        # Após verificar todas as temperaturas:
        # Se a flag 'fora_do_limite' ainda for Falsa (ou seja, NENHUMA temperatura
        # estava fora do limite):
        if not fora_do_limite:
            # Imprime uma mensagem de segurança.
            print("✅ Todas as temperaturas estão dentro do limite seguro.")



# 6 - Demonstração do uso do sistema

# Criação de um objeto (instância) da classe MonitorDeTemperatura.
# O método __init__ é chamado aqui.
monitor = MonitorDeTemperatura()

# Lista de temperaturas simuladas para teste.
leituras = [15, 30, 90, 45, 70] # Contém 15 (abaixo de 20) e 90 (acima de 80)

# Adiciona cada leitura ao monitor
for temp in leituras:
    # Chama o método 'adicionar_leitura' para cada temperatura na lista.
    monitor.adicionar_leitura(temp)

# Calcular e exibir as estatísticas
# Chama o método 'calcular_estatisticas' no objeto 'monitor'.
estatisticas = monitor.calcular_estatisticas()

# Exibição formatada dos resultados.
print("\n📊 Estatísticas das temperaturas registradas:")
print(f"Temperatura Máxima: {estatisticas['máxima']} °C")
print(f"Temperatura Mínima: {estatisticas['mínima']} °C")
# Usa formatação ': .2f' para exibir a média com duas casas decimais.
print(f"Temperatura Média: {estatisticas['média']:.2f} °C")

# Verifica se estão dentro do limite seguro
print("\n🔍 Verificando segurança das temperaturas:")
# Chama o método 'verificar_seguranca', que imprimirá alertas
# para 15°C e 90°C, e como há alertas, a mensagem final de '✅ seguro' não é exibida.
monitor.verificar_seguranca()
```
